## Ejercicio 1. Validador y clasificador de triángulos

```python
lado1 = float(input("Ingrese el primer lado: "))
lado2 = float(input("Ingrese el segundo lado: "))
lado3 = float(input("Ingrese el tercer lado: "))

if lado1 <= 0 or lado2 <= 0 or lado3 <= 0:
    print("Error: todos los lados deben ser mayores que cero.")
elif lado1 + lado2 <= lado3 or lado1 + lado3 <= lado2 or lado2 + lado3 <= lado1:
    print("Los lados ingresados no forman un triángulo.")
else:
    print("Los lados ingresados forman un triángulo válido.")
    if lado1 == lado2 == lado3:
        print("Tipo de triángulo: equilátero.")
    elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
        print("Tipo de triángulo: isósceles.")
    else:
        print("Tipo de triángulo: escaleno.")
```

## Ejercicio 2. Registro y análisis de números

```python
cantidad = int(input("¿Cuántos números desea analizar?: "))
if cantidad <= 0:
    print("Error: la cantidad debe ser mayor que cero.")
else:
    positivos = 0
    negativos = 0
    ceros = 0
    suma = 0
    for i in range(1, cantidad + 1):
        numero = float(input(f"Ingrese el número {i}: "))
        suma += numero
        if numero > 0:
            positivos += 1
        elif numero < 0:
            negativos += 1
        else:
            ceros += 1
    print(f"Cantidad de positivos: {positivos}")
    print(f"Cantidad de negativos: {negativos}")
    print(f"Cantidad de ceros: {ceros}")
    print(f"Suma total: {suma:g}")
    print(f"Promedio: {suma / cantidad:.2f}")
```

## Ejercicio 3. Menú de cálculos geométricos

```python
import math

while True:
    print("\n--- CALCULADORA GEOMÉTRICA ---")
    print("1. Área de un círculo")
    print("2. Calcular hipotenusa")
    print("3. Resolver ecuación cuadrática")
    print("4. Salir")
    opcion = input("Seleccione una opción: ")
    if opcion == "1":
        radio = float(input("Ingrese el radio: "))
        if radio < 0:
            print("Error: el radio no puede ser negativo.")
        else:
            print(f"Área del círculo: {math.pi * radio ** 2:.2f}")
    elif opcion == "2":
        a = float(input("Ingrese el primer cateto: "))
        b = float(input("Ingrese el segundo cateto: "))
        if a <= 0 or b <= 0:
            print("Error: los catetos deben ser mayores que cero.")
        else:
            print(f"Hipotenusa: {math.sqrt(a ** 2 + b ** 2):.2f}")
    elif opcion == "3":
        a = float(input("Ingrese el coeficiente a: "))
        b = float(input("Ingrese el coeficiente b: "))
        c = float(input("Ingrese el coeficiente c: "))
        if a == 0:
            print("No es una ecuación cuadrática.")
        else:
            discriminante = b ** 2 - 4 * a * c
            if discriminante > 0:
                raiz = math.sqrt(discriminante)
                print(f"Primera solución: {(-b + raiz) / (2 * a):.2f}")
                print(f"Segunda solución: {(-b - raiz) / (2 * a):.2f}")
            elif discriminante == 0:
                print(f"Única solución: {-b / (2 * a):.2f}")
            else:
                print("No existen soluciones reales.")
    elif opcion == "4":
        print("Programa finalizado.")
        break
    else:
        print("Opción inválida. Intente de nuevo.")
```

## Ejercicio 4. Analizador de contraseñas

```python
contrasena = input("Ingrese una contraseña: ")
mayuscula = False
minuscula = False
numero = False
espacios = False

for caracter in contrasena:
    if caracter.isupper():
        mayuscula = True
    if caracter.islower():
        minuscula = True
    if caracter.isdigit():
        numero = True
    if caracter.isspace():
        espacios = True

longitud = len(contrasena)
criterios = int(longitud >= 8) + int(mayuscula) + int(minuscula) + int(numero)
if espacios:
    clasificacion = "Débil"
else:
    if criterios == 4:
        clasificacion = "Fuerte"
    elif criterios == 3:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Débil"

print(f"Longitud: {longitud}")
print(f"Contiene mayúscula: {'Sí' if mayuscula else 'No'}")
print(f"Contiene minúscula: {'Sí' if minuscula else 'No'}")
print(f"Contiene número: {'Sí' if numero else 'No'}")
print(f"Contiene espacios: {'Sí' if espacios else 'No'}")
print(f"Criterios cumplidos: {criterios} de 4")
print(f"Clasificación: {clasificacion}")
```
